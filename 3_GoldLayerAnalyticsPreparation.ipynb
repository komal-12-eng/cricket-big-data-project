```python
import requests
import json
import pandas as pd
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
silver_df = spark.table('workspace.default.cricket_silver_current_matches')

silver_df.limit(5).display()
```

```python
gold_match_type = silver_df.groupBy('match_type').agg(count("*").alias('Total_Matches'))
display(gold_match_type)
```

```python
gold_venue_df = silver_df.groupBy('venue').agg(count("*").alias('Total_Matches')).orderBy('Total_Matches', ascending=False)
display(gold_venue_df)
```

```python
team1_df = silver_df.select(col('team_1').alias('Team'))
team2_df = silver_df.select(col('team_2').alias('Team'))
team_df = team1_df.union(team2_df)
team_df = team_df.groupBy('Team').agg(count("*").alias('Total_Matches')).orderBy(col('Total_Matches').desc())
display(team_df)

```

```python
spark.sql("""
          select count(*) as Total_matches,
          count(distinct match_type) as total_match_types,
          count(distinct venue) as total_venues
           from workspace.default.cricket_silver_current_matches"""
           ).display()
```
